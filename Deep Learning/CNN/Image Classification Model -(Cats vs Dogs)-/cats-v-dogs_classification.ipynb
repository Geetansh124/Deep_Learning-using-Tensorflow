```python
import sys
import subprocess
exe = sys.executable
for pkg in ['tensorflow', 'tensorflow-gpu', 'keras', 'h5py', 'numpy', 'scipy']:
    try:
        out = subprocess.check_output([exe, '-m', 'pip', 'show', pkg], text=True)
        print(f'=== {pkg} ===')
        print(out.strip())
    except subprocess.CalledProcessError:
        print(f'=== {pkg} not installed ===')
    except Exception as err:
        print(f'Error checking {pkg}:', err)
```

```python
import os
print('cwd:', os.getcwd())
print('root dirs:', sorted(os.listdir('.')))
for path in ['train','test']:
    if os.path.exists(path):
        print(path, 'exists')
        entries = sorted(os.listdir(path))
        print('  entries count:', len(entries))
        for entry in entries[:10]:
            full = os.path.join(path, entry)
            print('   ', entry, 'dir' if os.path.isdir(full) else 'file', 'items=' + str(len(os.listdir(full))) if os.path.isdir(full) else '')
        if len(entries) > 10:
            print('   ... plus', len(entries)-10, 'more entries')
    else:
        print(path, 'missing')
```
